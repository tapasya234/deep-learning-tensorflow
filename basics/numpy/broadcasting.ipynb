# NumPy Basics: Broadcasting

```python
import numpy as np
```

```python
A = np.array([2, 4, 6])
```

```python
# Add value 3 to every element in the list
A + 3
```

```python
A + np.array([3])
```

```python
# To broadcast, the elements shapes should match

# If they don't, an error is thrown
A + np.array([2, 3])
```

```python
2 * A
```

```python
# Broadcasting when the inner dimensions of the two matrices are the same
A = np.arange(3).reshape((3, 1))
B = np.arange(2).reshape((1, 2))
print(A)
print(A.shape)

print(B)
print(B.shape)
```

```python
A + B
```

#### Explanation for the above broadcasting summation

A is of size `m x n` and B is of size `n x o`. Since the inner dimentions are the same, before the summation, A is updated to be of size `m x o` and B is updated to be of size `m x o` and they are summed. 

$$
A = \begin{bmatrix}
0 \\
1 \\
2
\end{bmatrix}
\textrm {is updated to }
\begin{bmatrix}
0 & 0 \\
1 & 1 \\
2 & 2
\end{bmatrix}
$$

and 

$$
B = \begin{bmatrix}
0 & 1 
\end{bmatrix}
\textrm{is updated to }
\begin{bmatrix}
0 & 1 \\
0 & 1 \\
0 & 1 
\end{bmatrix}
$$
