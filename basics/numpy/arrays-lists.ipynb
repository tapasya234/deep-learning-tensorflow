## NumPy Basics: Arrays vs Lists

```python
import numpy as np
```

```python
# Python List
L = [1, 2, 3]

# NumPy array
A = np.array([1, 2, 3])
```

```python
# Iteration
print("Python List elements:")
for e in L:
    print(e)

print("NumPy Array elements:")
for e in A:
    print(e)
```

```python
# Appending to the list
L.append(12)
print(L)

# Appending to the array - Not possible because the size of the array is fixed.
try:
    A.append(123)
except Exception as exception:
    print("EXCEPTION: ", exception)
```

```python
# Can also append using the `+` operator
L + [5]
```

```python
# Operations on Arrays vs Lists

# Multiplication
5 * A
```

```python
# On Python lists, it'll concatenate the same list that many times
5 * L
```

```python
# Sum
A + A
```

```python
L + L
```

```python
# The same operation can be achieved in NumPy using a for loop
L2 = [e * 5 for e in L]
L2
```

```python
# Can also apply various methods
np.exp(A)
```
