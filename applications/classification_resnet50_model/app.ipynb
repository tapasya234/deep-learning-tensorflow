# Image Classification using Pre-trained ResNet50

````python
import tensorflow as tf
import keras
import matplotlib.pyplot as plt
import glob
import ssl
````

### Read images

````python
image_paths = glob.glob("input/*.jpg")
````

### Load the model

The ResNet50 model has been pre-trained on the ImageNet 1K dataset, which contains more than 1.2 million training images distributed across 1,000 classes. Such pre-trained models can be used out-of-the-box for many types of classification problems as long as the images intended to classify are represented in the ImageNet dataset. 

The Keras API provides the following utility to instantiate a ResNet50 model. 
```
tf.keras.applications.resnet50.ResNet50(include_top=True, 
                                        weights='imagenet', 
                                        input_tensor=None,
                                        input_shape=None, 
                                        pooling=None, 
                                        classes=1000,
                                        classifier_activation='softmax',
                                       )
```

````python
ssl._create_default_https_context = ssl._create_unverified_context
````

````python
model = tf.keras.applications.ResNet50(weights="imagenet")
print(model.output_shape)
````

## Perform Inference
Perform inference using Keras API to make predictions by following the below steps:

- Read the image
- Perform the required pre-processing as required by the model. For example - resizing the images to a standard size of 224 x 224.
- Create a data batch from the images. This is needed because the predict method works on image batches.
- Call the predict() method.
- Decode the predictions to find the class name and confidence score of predictions.

````python
for path in image_paths:
    imageEncoded = tf.io.read_file(path)

    # Decode the image by converting the Bytes string into a Tensor
    imgTF = tf.image.decode_image(imageEncoded)

    imgTF = tf.image.resize(imgTF, (224, 224))
    # Add batch dimension at the beginning
    imgTF = tf.expand_dims(imgTF, axis=0)
    # Preprocess the batched image to bring it to scale
    imgTF = keras.applications.imagenet_utils.preprocess_input(imgTF)

    predictions = model.predict(imgTF)

    # Decod the predictions from class number to class name.
    predictions = keras.applications.imagenet_utils.decode_predictions(
        predictions, top=5
    )

    # Display the image using plt
    imgPLT = plt.imread(path)
    plt.imshow(imgPLT)
    plt.axis("off")
    plt.show()

    for num, pred in enumerate(predictions[0]):
        print(f"Prediction {num}: {pred[1]}, {pred[2]*100:.2f}%")
````
