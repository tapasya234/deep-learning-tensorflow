## List of all Pre-trained Models in Keras

TensorFlow provides a host of pre-trained models through its `tf.keras.applications` API. It contains image classification models that have been pre-trained on the ImageNet dataset.

```python
import keras
```

```python
all_modules = dir(keras.applications)

# Navigate the `keras.applications` namespace and display the available models.
for module in all_modules:
    if module[0].islower():
        if module != "imagenet_utils":
            # Ignore ImageNet utils
            print(f"Model Family: {module}")
        if module == "mobilenet_v3":
            # Handle special case for mobilenet_v3.
            temp = "MobileNetV3Large"
            print(f"\t  |__ {temp}")
            temp = "MobileNetV3LSmall"
            print(f"\t  |__ {temp}")
        else:
            # Handle general cases.
            models = dir(getattr(keras.applications, module))
            for model in models:
                if model[0].isupper():
                    print(f"\t  |__ {model}")
        print()
```
